```python

marks = [45, 32, 78, 39, 90, 28, 55, 40, 67, 12]

pass_count = 0
fail_count = 0

for mark in marks:
    if mark >= 40:
        print(mark, "Pass")
        pass_count += 1
    else:
        print(mark, "Fail")
        fail_count += 1

passing_marks = [mark for mark in marks if mark >= 40]

print("\nTotal passed:", pass_count)
print("Total failed:", fail_count)
print("Passing marks:", passing_marks)
```

```python

def calculate_average(numbers: list) -> float:
    """Calculate and return the average of a list of numbers."""
    return sum(numbers) / len(numbers)

try:
    result = calculate_average([10, 20, 30])
    print("Average:", result)
except ZeroDivisionError:
    print("Error: Cannot calculate the average of an empty list.")

try:
    result = calculate_average([])
    print("Average:", result)
except ZeroDivisionError:
    print("Error: Cannot calculate the average of an empty list.")
```

```python

import random
import math

steps = []

for day in range(1, 8):
    daily_steps = random.randint(2000, 12000)
    steps.append(daily_steps)
    print(f"Day {day}: {daily_steps} steps")

print("\nHighest steps:", max(steps))
print("Lowest steps:", min(steps))

average_steps = math.ceil(sum(steps) / len(steps))
print("Average steps (rounded up):", average_steps)
```

```python

class Student:
    def __init__(self, name: str, marks: list) -> None:
        self.name: str = name
        self.marks: list = marks

    def average(self) -> float:
        return sum(self.marks) / len(self.marks)

    def __str__(self) -> str:
        return f"{self.name} average {self.average():.1f}"

students = [
    Student("Sakshi", [80, 85, 82]),
    Student("Rahul", [70, 75, 80]),
    Student("Priya", [90, 95, 92])
]

for student in students:
    print(student)

top_student = max(students, key=lambda student: student.average())

print("\nStudent with the highest average:", top_student.name)
print("Highest average:", top_student.average())
```

```python

import math

radius = float(input("Enter the radius of the circle: "))

radius_squared = math.pow(radius, 2)

area = math.pi * radius_squared

rounded_area = math.ceil(area)

circumference = 2 * math.pi * radius

print("Area:", area)
print("Rounded area:", rounded_area)
print("Circumference:", circumference)
```
